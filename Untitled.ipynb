```python
import numpy as np
import matplotlib.pyplot as plt
import librosa

def plot_mp3_global_dft_librosa(mp3_path):
    # 1. Load the MP3 audio file
    print("Loading MP3 file with librosa...")
    # librosa.load automatically converts stereo to mono and normalizes the data
    # sr=None preserves the original sample rate of the file
    data, sample_rate = librosa.load(mp3_path, sr=None, mono=True)
    
    # 2. Compute the Global DFT
    print(f"Computing global DFT for {len(data)} samples...")
    # Using np.fft.rfft is highly optimized for real-valued inputs (like audio)
    # It automatically returns only the positive frequencies, saving memory.
    dft_output = np.fft.rfft(data)
    
    # Extract the magnitude spectrum
    magnitude = np.abs(dft_output)
    
    # Generate mapping for frequency bins
    # rfftfreq directly matches the output of rfft
    frequencies = np.fft.rfftfreq(len(data), d=1/sample_rate)
    
    # 3. Visualization
    plt.figure(figsize=(12, 6))
    
    # Convert magnitude to decibel (dB) scale for dynamic range visualization
    magnitude_db = 20 * np.log10(magnitude + 1e-5)
    
    # Plot frequency profile
    plt.plot(frequencies, magnitude_db, color='crimson', linewidth=0.5)
    plt.title("Global DFT Magnitude Spectrum (via Librosa)\nTiming Information is Completely Obliterated", fontsize=13)
    plt.xlabel("Frequency (Hz)", fontsize=11)
    plt.ylabel("Magnitude (dB)", fontsize=11)
    plt.grid(True, linestyle='--', alpha=0.5)
    
    # Focus on 0 to 5000 Hz where musical audio is dense
    plt.xlim(0, 5000)  
    
    plt.tight_layout()
    plt.show()

# Example execution:
plot_mp3_global_dft_librosa('./song data/A Day In The Life.mp3')
```
